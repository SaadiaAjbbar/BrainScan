```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input,
    Conv2D,
    MaxPooling2D,
    Dropout,
    GlobalAveragePooling2D,
    Dense
)
```

```python
IMG_HEIGHT = 224
IMG_WIDTH = 224
CHANNELS = 3
NUM_CLASSES = 4

print("Hauteur :", IMG_HEIGHT)
print("Largeur :", IMG_WIDTH)
print("Canaux :", CHANNELS)
print("Nombre de classes :", NUM_CLASSES)
```

```python
model = Sequential([
    # Définir la forme de l'image en entrée
    Input(shape=(IMG_HEIGHT, IMG_WIDTH, CHANNELS)),

    # Bloc convolutionnel 1
    Conv2D(32, kernel_size=(3, 3), activation="relu", padding="same"),
    MaxPooling2D(pool_size=(2, 2)),

    # Bloc convolutionnel 2
    Conv2D(64, kernel_size=(3, 3), activation="relu", padding="same"),
    MaxPooling2D(pool_size=(2, 2)),

    # Bloc convolutionnel 3
    Conv2D(128, kernel_size=(3, 3), activation="relu", padding="same"),
    MaxPooling2D(pool_size=(2, 2)),

    # Réduire le nombre de paramètres
    GlobalAveragePooling2D(),

    # Couche dense de classification
    Dense(128, activation="relu"),

    # Régularisation
    Dropout(0.4),

    # Couche de sortie : 4 classes
    Dense(NUM_CLASSES, activation="softmax")
])

print("Architecture CNN créée avec succès.")
```

```python
print("Type du modèle :", type(model).__name__)
print("Nombre de couches :", len(model.layers))
```

```python
HIDDEN_ACTIVATION = "relu"
OUTPUT_ACTIVATION = "softmax"

print("Fonction d'activation des couches cachées :", HIDDEN_ACTIVATION)
print("Fonction d'activation de la couche de sortie :", OUTPUT_ACTIVATION)
```

```python
model = Sequential([
    Input(shape=(IMG_HEIGHT, IMG_WIDTH, CHANNELS)),

    Conv2D(
        32,
        kernel_size=(3, 3),
        activation=HIDDEN_ACTIVATION,
        padding="same"
    ),
    MaxPooling2D(pool_size=(2, 2)),

    Conv2D(
        64,
        kernel_size=(3, 3),
        activation=HIDDEN_ACTIVATION,
        padding="same"
    ),
    MaxPooling2D(pool_size=(2, 2)),

    Conv2D(
        128,
        kernel_size=(3, 3),
        activation=HIDDEN_ACTIVATION,
        padding="same"
    ),
    MaxPooling2D(pool_size=(2, 2)),

    GlobalAveragePooling2D(),

    Dense(128, activation=HIDDEN_ACTIVATION),

    Dropout(0.4),

    Dense(NUM_CLASSES, activation=OUTPUT_ACTIVATION)
])

print("Architecture CNN créée avec les fonctions d'activation choisies.")
```

```python
model.summary()
for layer in model.layers:
    if hasattr(layer, "activation"):
        print(
            layer.name,
            "->",
            layer.activation.__name__
        )
```

```python
LEARNING_RATE = 0.001
EPOCHS = 20
BATCH_SIZE = 32

print("Learning rate :", LEARNING_RATE)
print("Nombre d'époques :", EPOCHS)
print("Batch size :", BATCH_SIZE)
```

```python
from tensorflow.keras.optimizers import Adam

model.compile(
    optimizer=Adam(learning_rate=LEARNING_RATE),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

print("Modèle compilé avec succès.")
```
